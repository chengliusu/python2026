# 修改值

```python
import numpy as np
 
t = np.arange(24).reshape(4, 6)
print(t)
print(id(t))
print('-' * 50)
```

```python
# 修改某一行的值，支持批量修改
t[1, :] = 0

# 修改某一列的值
t[:, 1] = 0

# 修改连续多行
t[1:3, :] = 0

# 修改连续多列
t[:, 1:4] = 0

# 修改多行多列，取下标1-2行，下标2-4列
t[1:3, 2:5] = 0

# 修改多个不相邻的点
t[[0, 1], [1, 3]] = 0       # t[0,1] = 0, t[1, 3] = 0

# 可以根据条件修改，比如讲小于10的值改掉
t[t<10] = 0

# python中 & 是位运算符，作用于二进制位，and 是逻辑运算符，判断真假
# 使用逻辑判断
# np.logical_and, np.logical_or, np.logical_not
# numpy中布尔条件必须用 & | ~ (位运算符)，不能直接用and 、or
t[(t > 2) & (t < 6)] = 0        # 逻辑与，and
t[(t < 2) | (t > 6)] = 0        # 逻辑或，or
t[~(t > 6)] = 0                 # 逻辑非
print(t)

# 对numpy数组t执行‘裁剪’操作-将数组中小于10的元素替换为10，大于18的元素替换为18，介于10-18之间的元素保持不变，最后将裁剪后的数组重新赋值给t
t = t.clip(10, 18)
print(id(t))
t
```

# 数据的添加、删除与去重

```python
import numpy as np

a = np.array([[1, 2, 3], [4, 5, 6]])
# np.append(arr, values, axis=None) 默认axis=None：先展平，再追加
print(np.append(a, [7, 8, 9]))
print('-' * 50)

# ValueError: all the input arrays must have same number of dimensions, but the array at index 0 has 2 dimension(s) and the array at index 1 has 1 dimension(s)
# print(np.append(a, [7, 8, 9], axis=0))

# 这里的values必须与a维度相同
# 行添加
print(np.append(a, [[7, 8, 9]], axis=0))
print('-' * 50)
# 列添加
print(np.append(a, [[5, 5, 5], [7, 8, 9]], axis=1))
print('-' * 50)
# 所有append操作返回新数组，原来数组不发生变化
print(a)
```

```python
b = np.array([[7, 8, 9]])
print(b.ndim)
print(b)
```

```python
# np.insert(arr, obj, values, axis=None)    arr：原数组；obj：插入位置索引（下标）；values：要插入的数据；默认axis=None：先把数组展平成一维再插入
a = np.array([[1, 2], [3, 4], [5, 6]])
# 默认展平插入
print(np.insert(a, 1, [11, 22]))
print('-' * 50)
# 按行插入（insert可以一维，但是append必须二维）
print(np.insert(a, 1, [11, 22], axis=0))
print('-' * 50)
# 按列插入
print(np.insert(a, 1, [33, 44, 55], axis=1))
print('-' * 50)
# 如果只有一个元素值11，则会将11在0轴复制展开
print(np.insert(a, 1, 11, axis=0))
print('-' * 50)
# 所有插入操作都是返回新数组，原数组不发生变化
print(a)
```

```python
# np.delete(arr, obj, axis=None)    arr：原数组；obj：要删除的下标/下标数组；默认axis=None：先把数组扁平化（拉成一维）再删除
import numpy as np
a = np.arange(12).reshape(3, 4)
print(a)
print('-' * 50)
# 默认展平删除，删除下标为5的数据
print(np.delete(a, 5))
print('-' * 50)
# 删除下标为1的行
print(np.delete(a, 1, axis=0))
print('-' * 50)
# 删除下标为1的列
print(np.delete(a, 1, axis=1))
print('-' * 50)
# 所有删除操作都是返回新数组，原数组不发生变化
print(a)
```

```python
# np.unique(a) 去重，返回数组中不重复的元素，并默认排序 
import numpy as np
a = np.array([5, 2, 6, 2, 7, 5, 6, 9, 8, 2])

# 返回新数组，原数组不发生变化
print(np.unique(a))
print('-' * 50)
# np.unique(arr, return_index=False, return_inverse=False, return_counts=False, axis=None)
# return_index=True：返回唯一元素 + 首次出现下标；vals, idx = np.unique(a, return_index=True)
u, idx = np.unique(a, return_index=True)
print(u)
print(idx)
print('-' * 50)
# return_counts=True：返回唯一元素 + 每个元素出现次数；vals, cnt = np.unique(a, return_counts=True)
u, cnt = np.unique(a, return_counts=True)
print(u)
print(cnt)
print('-' * 50)
# return_inverse=True：返回唯一元素 + 映射下标（可还原原数组）；vals, inv = np.unique(a, return_inverse=True)   recover = vals[inv] 还原原始数组
u, inv = np.unique(a, return_inverse=True)
print(u)                            # [2 5 6 7 8 9]
print(inv)                      # [1 0 2 0 3 1 2 5 4 0]     这里的数据是原数据 a 在 u 中的下标
```

# numpy的数学运算

```python
import numpy as np
score = np.array([[80, 88], [82, 81], [75, 81]])
print(score)
print(score.shape)
print('-' * 50)

# 1. 获取所有数据最大值
result = np.max(score)
print(result)
print('-' * 50)

# 2. 获取某一个轴上的数据最大值，对应的轴就消失
# 每行的各列求最大值
result = np.max(score, axis=1)
print(result)           # [88 82 81]
print('-' * 50)

# 3. 获取所有数据最小值
result = np.min(score)
print(result)
print('-' * 50)

# 4. 获取某一个轴上的数据最小值
# 各列的每行求最小值
result = np.min(score, axis=0)
print(result)           # [75 81]
print('-' * 50)

# 5. 数据的比较
# np.maximum(x1, x2)    逐元素进行比较，取两者中较大的值
result = np.maximum([-2, -1, 0, 1, 2], 0)
print(result)
# np.minimum(x1, x2)    逐元素进行比较，取两者中较小值
result = np.minimum([-2, -1, 0, 1, 2], 0)
print(result)
# 对应元素进行比较，取两者较大值；接受的两个参数可以大小一致，当第二个参数x2只是一个单独的值时，其实用到了维度的广播机制
result = np.maximum([-2, -1, 4, 1, 2], [1, 2, 3, 4, 5])
print(result)
```

```python
# 6. 求平均值
# 获取所有数据的平均值
result = np.mean(score)
print(result)
print('-' * 50)

# 各列的每行求均值
result = np.mean(score, axis=0)
print(result)
print('-' * 50)

# 7. 求前缀和
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr)
print('-' * 50)

# cumsum = cumulative sum累加和（前缀和）
'''
[1, 2, 3]------>	|1	   |2	  |3	|
[4, 5, 6]------>	|5=1+4 |7=2+5 |9=3+6|
'''
print(arr.cumsum(axis=0))
print('-' * 50)
'''
[1, 2, 3]------>	|1	|2+1	|3+2+1	|
[4, 5, 6]------>	|4	|4+5	|4+5+6	|
'''
print(arr.cumsum(axis=1))

# 拓展：方差var, 协方差cov, 计算平均值 average, 计算中位数 median
```

```python
print(score)
print('-' * 50)
# 8. argmin求最小值索引（argument of minimum：找出最小值所在的下标）
# 每列的各行中最小值索引
result = np.argmin(score, axis=0)
# 每列的各行中的最小值
res = np.min(score, axis=0)
print(result, res)
print('-' * 50)

# 9. 求每一列标准差（这里是总体标准差 standard deviation）
# 标准差是一组数据平均值分散程度的一种度量。一个较大的标准差，代表大部分数值和其平均值之间差异较大。
# 一个较小的标准差，代表这些数据较接近平均值反映出数据的波动稳定情况，越大表示波动越大，越不稳定。
result = np.std(score, axis=0)
print(result)
print('-' * 50)

# 10. 极值(peak to peak：最大值与最小值的差)
result = np.ptp(score, axis=None)
print(result)
```

# 数组的拼接

```python
# 有时候我们需要将两个数据加起来一起研究分析，我们就可以将其进行拼接然后分析
import numpy as np

# 1. 根据轴连接的数组序列，concatenate没有改变数组的维度
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
c = np.array([[9, 10], [11, 12]])
# 要求a，b两个数组的维度相同
print('沿轴 0 连接两个数组：')
print(np.concatenate((a, b), axis=0))
```

```python
print('沿轴 1 连接三个数组：')
print(np.concatenate((a, b, c), axis=1))
```

# numpy的 inf 和 nan

```python
# inf: infinity  Nan: Not a number（表示非数字，在numpy中表示一个非数字值，可以用来表示缺失数据）
import numpy as np
a = np.nan
b = np.inf
print(a, type(a))
print(b, type(b))
```

```python
# 并且 np.nan != np.nan   结果是 True
# 所以我们可以使用这两个结合使用判断nan的个数
print(np.nan == np.nan)
print(True == 1)
print(False == 0)
np.nan + 1      # nan和其他数运算的结果都是nan
```

```python
# 判断数组中nan的个数
t = np.arange(24, dtype=float).reshape(4, 6)

# 将数组中行索引为3、列索引为4 这个位置的元素设置为np.nan(Not a Number，非数值，即缺失值、无效值)
t[3, 4] = np.nan
t[2, 4] = np.nan
print(t)
```

```python
# 可以使用 np.count_nonzero() 来判断非零个数
print(np.count_nonzero(t))
print(t != t)
print('-' * 50)
# 统计nan个数
print(np.count_nonzero(t != t))
```

```python
# 生成和 t 同形状的布尔数组：元素NaN -> True，正常数字 -> False
print(np.isnan(t))
# 布尔索引，选中所有NaN位置，把所有NaN改成0
t[np.isnan(t)] = 0
print(t)
```

```python
# ------------------------练习：处理数组中的nan
t = np.arange(24).reshape(4, 6).astype('float')

# 将数组中一部分替换nan
t[1, 3:] = np.nan
print(t)
print('-' * 50)

print(t.shape)              # (4, 6)
print(type(t.shape))        # <class 'tuple'>
print(id(t))
print('-' * 50)

# 遍历每一列，然后判断每一列是否有nan
# t.shape[0] = 行数；t.shape[1] = 列数   t.shape = (4, 6)    t.shape[0] = 4     t.shape[1] = 6
for i in range(t.shape[1]):
    # 获取当前列数据
    temp_col = t[:, i]
    # 判断当前列的数据中是否含有nan
    nan_num = np.count_nonzero(temp_col != temp_col)
    # 条件成立说明含有nan
    if nan_num > 0:
        # 将这一列不为nan的数据拿出来，并计算平均值
        temp_col_not_nan = temp_col[temp_col == temp_col]
        print(temp_col_not_nan)
        # 将nan替换成这一列的平均值（np.isnan(temp_col) -> 同形状布尔型数组）
        temp_col[np.isnan(temp_col)] = np.mean(temp_col_not_nan)
        
        
print(t)
print(id(t))
```

# 转置和轴滚动

```python
# 对换数组的维度
import numpy as np

a = np.arange(12).reshape(3, 4)
print(a)
print('-' * 50)

# 就是线性代数里的转置
print(np.transpose(a))
# 这里同样，转置是生成新的数组，而原数组不发生任何变化
print(a)
print('-' * 50)

# 与transpose一致
a = np.arange(12).reshape(3, 4)
print(a)
print(a.T)
```

```python
# 函数用于交换数组的两个轴
t1 = np.arange(24).reshape(4, 6)
print(t1)
print('-' * 50)

# axis（单数）-> axes（复数）
# swapaxes只能交换两个轴，而transpose可以交换任意多个轴
re1 = t1.swapaxes(1, 0)
print(re1)
print(re1.shape)
```

```python
t3 = np.arange(60).reshape(3, 4, 5)
print(t3.shape)
print('-' * 50)
# 0轴不变，1轴与2轴相互交换
t3 = np.swapaxes(t3, 1, 2)
print(t3.shape)
print(t3)
```

```python
# 数组的轴滚动，swapaxes每次只能交换两个轴，没有rollaxis方便，默认情况下轴滚动最前面
a = np.ones((3, 4, 5, 6))
# np.rollaxis(a, 2).shape
# 抽走轴3（其余轴相对顺序保持不变），将轴3放置在轴1的位置；故位置变为从 (0, 1, 2, 3) -> (0, 3, 1, 2)
np.rollaxis(a, 3, 1).shape
```

```python
# 数据拷贝，copy()和赋值的区别
b = np.array([[1, 2, 3], [1, 2, 3]])
a = b.copy()
print(id(b))
print(id(a))
a
```
