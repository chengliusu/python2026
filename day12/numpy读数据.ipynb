```python
import numpy as np
from matplotlib import pyplot as plt

us_file_path = "./youtube_video_data/US_video_data_numbers.csv"
uk_file_path = ".youtube_video_data/GB_video_data_numbers.csv"

# t1 = np.loadtxt(us_file_path, delimiter=",", dtype="int", unpack=True)
# us_file_path：文件路径（csv/txt文本文件）
# delimiter=","：分隔符，这里是逗号，对应csv文件
# dtype="int"：读取后数组元素类型为整数，不指定默认浮点
# unpack=True：默认False，按行读取，一行是数组一行；True：按列解包，每一列单独变成一维数组，按顺序打包返回
t_us = np.loadtxt(us_file_path, delimiter=",", dtype="int")

# 取美国的电影的评论数据
t_us_comments = t_us[:, -1]

# 选择比5000小的数据
# t_us_comments <= 5000：生成同形状布尔数组，元素 <= 5000 -> True，否则 False
# t_us_comments[布尔数组]：布尔索引筛选，取出条件为True的元素
# 重新赋值给变量：覆盖原数组，只保留 <= 5000 的数据
t_us_comments = t_us_comments[t_us_comments <= 5000]

# 以元组的形式表示，所以会显示一个逗号
print(t_us_comments.shape)      # (1373,)

# 再看看最小值和最大值
print(t_us_comments.max(), t_us_comments.min())
```

```python
# 剩余多少数据
print(t_us_comments.shape)
# 组距50
d = 50
# bin_nums表示组数
bin_nums = ((t_us_comments.max() - t_us_comments.min()) // d)

# 绘图
plt.figure(figsize=(20, 8), dpi=180)
plt.hist(t_us_comments, bins=bin_nums)

plt.grid(True)
plt.show()
```

```python
import numpy as np
from matplotlib import pyplot as plt

us_file_path = "./youtube_video_data/US_video_data_numbers.csv"
uk_file_path = "./youtube_video_data/GB_video_data_numbers.csv"

# t1 = np.loadtxt(us_file_path, delimiter=",", dtype="int", unpack=True)
# unpack实现转置的效果
t_uk = np.loadtxt(uk_file_path, delimiter=",", dtype="int")

# 取出来英国的YouTube评论数和喜欢数，第一列是like数，第三列是comment数
t_uk = t_uk[t_uk[:, 1] <= 250000]   # 筛选like数小于等于250000的视频
print(t_uk.shape)
t_uk_like = t_uk[:, 1]              # 取出like列，第一列，作为横坐标

t_uk_comment = t_uk[:, -1]          # 取吹评论数列，最后一列，作为纵轴坐标

plt.figure(figsize=(20, 8), dpi=180)
# 打印横轴纵轴标签
plt.xlabel("like", fontsize=20)
plt.ylabel("comment", fontsize=20)
plt.scatter(t_uk_like, t_uk_comment)    # 画散点图，观察横轴和纵轴的关系

plt.grid(True)
plt.show()
```
