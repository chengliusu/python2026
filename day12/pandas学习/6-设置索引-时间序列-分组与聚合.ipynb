# 9.Pandas分组与聚合
设置索引

```python
import pandas as pd
import numpy as np

a = pd.DataFrame({'a': range(7), 
                  'b': range(7, 0, -1), 
                  'c': ['one', 'one', 'one', 'two', 'two', 'two', 'two'],
                  'd': list('hjklmno')})
a
```

```python
c = a.copy()
print(c)
print('-' * 50)
a.index = list('abcdefg')                   # 更换a的索引
a.columns = list('ABCD')                    # 更换a的列名
print(a)
```

```python
c.values.shape
```

```python
print(a)
print('-' * 50)
# 将列C提升为行索引；inplace=True：直接原地修改原数据a，不返回新DataFrame
a.set_index(['C'], inplace=True)
a
```

# 时间序列

```python
pd.date_range(start='20260801', end='20260901')             # 左闭右闭
```

```python
# periods：指定生成多少个时间点    B：business day（工作日）
pd.date_range(start='20260804', periods=10, freq='B')
```

```python
# periods：生成10个时间点；ME：month end
pd.date_range(start='20260804', periods=10, freq='ME')
```

```python
pd.date_range(start='20260804', periods=10, freq='MS')                      # month start
```

```python
pd.date_range(start='20260804', periods=10, freq='W')                       # W：week；拿每周的周日生成
```

```python
s = pd.Series(['3/11/2000', '3/12/2000', '3/13/2000'] * 5)
print(len(s))
s
```

```python
pd.to_datetime(s)                   # to_datetime把字符串转为时间格式
```

```python
from matplotlib import pyplot as plt

# 把时间字符串转为时间类型设置为索引
df = pd.read_csv('./911.csv')
df['timeStamp'] = pd.to_datetime(df['timeStamp'])

temp_list = df['title'].str.split(':').tolist()                 # 二维列表
# 遍历temp_list里每一个元素i，取出i的第0号元素，拼成新列表
cate_list = [i[0] for i in temp_list]
# print(cate_list)
# print(np.array(cate_list).reshape((df.shape[0], 1)))
df['cate'] = pd.DataFrame(np.array(cate_list).reshape((df.shape[0], 1)))
print(type(df['cate']))                                         # <class 'pandas.Series'>
df.set_index('timeStamp', inplace=True)                         # 设置索引，时间戳，inplace=True表示在原df上修改
df.head()
```
