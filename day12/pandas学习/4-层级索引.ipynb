# 层级索引（hierarchical indexing）(机器学习、深度学习不重要)

```python
import pandas as pd
import numpy as np

# MultiIndex是层级索引，索引类型的一种
index1 = pd.MultiIndex.from_arrays([['a', 'a', 'a', 'b', 'b', 'b', 'c', 'c', 'c', 'd', 'd', 'd'], [0, 1, 2, 0, 1, 2, 0, 1, 2, 0, 1, 2]], names=['cloth', 'size'])

ser_obj = pd.Series(np.random.randn(12), index=index1)
print(ser_obj)
print(type(ser_obj))                                # <class 'pandas.Series'>
print(type(ser_obj.index))                          # <class 'pandas.MultiIndex'>
print(ser_obj.index)
print(ser_obj.index.levels)                         # 层级索引的索引值
ser_obj.index.codes
```

```python
ser_obj
```

```python
# 层级索引如何取数据
print(ser_obj['c'])                             # 取出c的所有数据，取出的是series
print('-' * 50)
print(ser_obj.loc['a', 2])
print('-' * 50)
print(ser_obj[:, 2])                            # 取出所有行的内层索引为2的数据
```

# 交换层级

```python
print(ser_obj.swaplevel())
print('-' * 50)
print(ser_obj)
print('-' * 50)
ser_obj = ser_obj.swaplevel()
print(ser_obj)
```

```python
print(ser_obj.sort_index(level=0))              # 层级索引按索引级别排序，level=0表示最外层索引排序
```

```python
ser_obj                                         # 原数据不变
```

```python
# ser.unstack(level)    默认level=-1：展开最内层索引
# 展开内层索引作列
df_obj = ser_obj.unstack()
print(df_obj)
```

```python
print(df_obj)
# 把列索引放入内层
print(df_obj.stack())
```
