# 处理重复数据

```python
import pandas as pd
import numpy as np
```

```python
df_obj = pd.DataFrame({'data1': ['a'] * 4 + ['b'] * 4, 
                       'data2': np.random.randint(0, 4, 8)})
print(df_obj)
```

```python
# 判断每一行是否是重复行，返回一个布尔Series  True：该行是重复行（前面已经出现过一摸一样的行）；False：该行是首次出现，不重复
print(df_obj.duplicated())
```

```python
# 取出不重复的行
df_obj[~df_obj.duplicated()]
```

```python
# 按照某一列去重
print(df_obj.duplicated('data2'))
```

```python
df_obj1 = pd.DataFrame({'data1': [np.nan] * 4,
                        'data2': list('1234')})
df_obj1
```

```python
print(df_obj1.duplicated('data1'))
print(df_obj1.drop_duplicates('data1'))             # 后三行数据被删除
```

```python
print(df_obj.drop_duplicates())                     # 删除重复行，左右数据全部相同的行才删除
print(df_obj.drop_duplicates('data2'))              # 特定列数据相同的行删除
```

```python
# 如果要在原有的df上去重，需要加inplace=True
df_obj
```

```python
# map与applymap一样，但是map只能用于Series，applymap只能用于df
ser_obj = pd.Series(np.random.randint(0, 10, 10))       # Series用于map
print(ser_obj)

print(ser_obj.map(lambda x: x**2))
```

```python
ser_obj = pd.Series(np.arange(10), index=range(3, 13))

# 这里是值 1 替换成 -100
print(ser_obj.replace(1, -100))
print('-' * 50)

# 注意：这里range(6, 9)是索引的下标，将索引的下标的位置的数据换成-100
print(ser_obj.replace(range(6, 9), -100))
print('-' * 50)

# 将索引的下标4、7位置的数据分别换成-100、-200
print(ser_obj.replace([4, 7], [-100, -200]))
```

```python
df = pd.DataFrame({'A': [0, 1, 2, 3, 4],
                   'B': [5, 6, 7, 8, 9], 
                   'C': ['a', 'b', 'c', 'd', 'e']})
df
```

```python
# 使用正则表达式匹配单元格文本，匹配成功就替换成100
# r'^a'：正则，^代表字符串开头，匹配以字母a开头的字符串
# regex(regular expression)=True：开启正则模式（不加就只是普通字符串匹配，正则符号失效）
df.replace(to_replace=r'^a', value=100, regex=True)
```
