# Pandas的索引操作

```python
import pandas as pd
import numpy as np
```

```python
dict_data = {'A': 1,
             'B': pd.Timestamp('20190926'),
             'C': pd.Series(1, index=list(range(4)), dtype='float32'),
             'D': np.array([1, 2, 3, 4], dtype='int32'),
             'E': ['Python', 'Java', 'C++', 'C'],
             'F': 'wangdao'}
df_obj2 = pd.DataFrame(dict_data)
print(df_obj2)
print('-' * 50)
print(df_obj2.index)
```

```python
# df_obj2.index 是 index 对象（不可变索引序列）；df_obj2.index[0] = 2 索引对象的值不可变
print(df_obj2.index[2])         # 下标为2的index值为2，故输出为2
```

# 3. 常见的Index种类
index,索引    可以是各种类型
Index64Index，整数索引
MultiIndex，层级索引，难度较大
DatatimeIndex，时间戳类型

```python
ser_obj = pd.Series(range(5), index=list('abcde'))
print(ser_obj)
ser_obj.index
```

```python
# 返回索引下标 d 的值 3
# 用索引名来索引返回值
print(ser_obj['d'])
print(ser_obj.loc['d'])
# iloc -> index location：通过索引下标返回值
print(ser_obj.iloc[3])
```

```python
# 这里是返回下标为3的index值
print(ser_obj.index[3])
```

```python
# 切片索引
print(ser_obj.iloc[1: 3])       # 索引位置取数据，左闭右开
print('-' * 50)

# 注意：左闭右闭
print(ser_obj.loc['b': 'd'])    # 通过索引名取数据，左闭右闭
```

```python
# 不连续索引
print(ser_obj.iloc[[0, 2, 4]])
print(ser_obj.loc[['a', 'e']])
```

```python
# 布尔索引
ser_bool = ser_obj > 2
print(ser_obj)
print('-' * 50)
print(ser_bool)
```

```python
# 布尔索引
print(ser_obj[ser_bool])
# 取出值大于2的元素
print(ser_obj[ser_obj > 2])
```

# 4.4 DataFrame索引

```python
import numpy as np
df_obj = pd.DataFrame(np.random.randn(5, 4), columns=['a', 'b', 'c', 'd'])
print(df_obj.head())
```

```python
# 列索引
print(df_obj['a'])              # 返回Series类型
print('-' * 50)

print(df_obj[['a']])            # 返回DataFrame类型
print(type(df_obj[['a']]))      # 返回DataFrame类型
```

# loc标签索引（通过索引标签值获取数据）

```python
# 标签索引 loc，建议使用loc，效率更高
# Series
print(ser_obj)
print(ser_obj['b': 'd'])
print(ser_obj.loc['b': 'd'])        # 左闭右闭
```

```python
# DataFrame
df_obj = pd.DataFrame(np.random.randn(5, 4), columns=list('ABCD'), index=list('abcde'))
print(df_obj)
print('-' * 50)

print(df_obj['A'])                  # 建议不用，拿的是列
print('-' * 50)

print(df_obj.loc['a'])              # 拿的是行
```

```python
# 第一个参数索引行，第二个参数是列，loc或者iloc效率高于直接用取下标的方式，左闭右闭
# .loc[行切片, 列切片]，按标签选取；行：'a': 'c'，行标签从'a'到'c',包含首尾a、c（loc切片特点，和iloc左闭右开不一样）
print(df_obj.loc['a': 'c', 'A': 'C'])               # 连续索引
print('-' * 50)
print(df_obj.loc[['a', 'c'], ['B', 'D']])           # 不连续索引
print('-' * 50)
print(df_obj.loc[['c'], ['C']])                     # 取一个值，返回的是DataFrame类型
print('-' * 50)
print(df_obj.loc['c', 'C'])                         # 取一个值
```

# iloc 位置索引（推荐使用）

```python
print(ser_obj)
print('-' * 50)
# Series直接切片按位置，左闭右开，取下标1、2两个元素；等价ser_obj.iloc[1: 3]
# 直接通过索引的下标查找值
# ser_obj[1: 3] -> ser_obj.iloc[1: 3]
print(ser_obj[1: 3])
print('-' * 50)
print(ser_obj.iloc[1: 3])           # 左闭右开，效率高
```

```python
# DataFrame，iloc是左闭右开
print(df_obj)
print('-' * 50)
# .iloc[行切片，列切片]，纯位置索引，左闭右开
print(df_obj.iloc[0: 2, 0: 2])
print('-' * 50)
print(df_obj.iloc[[0, 2], [0, 2]])          # 不连续索引
print('-' * 50)
print(df_obj.iloc[0, 0])                    # 取一个值
```

```python
# 没有设置行和列索引的DataFrame，iloc和loc区别
df_obj2 = pd.DataFrame(np.random.randn(5, 4))
print(df_obj2)
print('-' * 50)
print(df_obj2.iloc[0: 2])                   # 左闭右开，两行
print('-' * 50)
print(df_obj2.loc[0: 2])                    # 这里是index索引，不是index索引下标；左闭右闭
```

# 5. 对齐运算

```python
import pandas as pd
s1 = pd.Series(range(10, 20), index=range(10))
print(s1)
print('-' * 50)
s2 = pd.Series(range(20, 25), index=range(5))
print(s2)
print('-' * 50)
# Series 对齐运算
print('s1 + s2: ')
s3 = s1 + s2
print(s3)                           # 缺失数据默认是Nan    np.nan
```

```python
# 两个长度不同的一维ndarray相加
a1 = np.array([1, 2, 3, 4, 5])
a2 = np.array([1])                  # 长度为1
print(a2.shape)
print(a1 + a2)
```

```python
print(s2)
s1
```

```python
print(np.isnan(s3[6]))
print('-' * 50)
print(s2.add(s1, fill_value=0))             # 未对齐数据将和填充值做运算 (s2 + s1)
print('-' * 50)
print(s2.sub(s1, fill_value=0))             # s2 - s1
```

```python
# df的对齐运算
import numpy as np
df1 = pd.DataFrame(np.ones((2, 2)), columns=['a', 'b'])
df2 = pd.DataFrame(np.ones((3, 3)), columns=['a', 'b', 'c'])
print(df1)
print(df2)
print('-' * 50)
print(df2.dtypes)
print(df1 - df2)
print(df2.sub(df1, fill_value=2))               # df2 - df1，df1中填充值为2，故1-2 = -1，填充部分运算后为 -1
```

# 总结：每对齐的元素，默认填充NaN，对齐运算时，fill_value参数可以指定填充值
