# Pandas的函数应用

```python
import pandas as pd
import numpy as np

# randn跟的是维数，5行4列
df = pd.DataFrame(np.random.randn(5, 4) - 1)
print(df)
print('-' * 50)

# absolute(绝对值)
print(np.abs(df))
```

```python
# apply默认作用在列上，因为axis=0，x依次代表每一列（Series）；x.max()求该列最大值；返回：每列最大值组成的Series，索引=列名
# df.apply(lambda x: x.max()) -> df.max()
# apply：DataFrame的核心方法，用于对指定轴（行/列）批量应用自定义函数
print(df.apply(lambda x: x.max()))
```

```python
"""
lambda x: x.max()
->
def func(x):
    return x.max()
"""
# apply作用在行上
print(df.apply(lambda x: x.max(), axis=1))
```

```python
print(df)
# round(x, 2)   对x四舍五入保留2位小数
print(df.map(lambda x: f'{x:.2f}'))
df.dtypes
```

```python
ser = pd.Series(['male','female','male'])
print(ser.map({'male':1, 'female':0}))
```

```python
type(f'{1.3456:.2f}')
```

# 6.4 索引排序

```python
# Series
# 在区间[0, 5)之间随机生成5个整数
print(np.random.randint(5, size=5))
print('-' * 50)
s4 = pd.Series(range(10, 15), index=np.random.randint(5, size=5))           # 索引随机生成
print(s4)
print('-' * 50)
# 索引排序，sort_index返回一个新的排好索引的series
print(s4.sort_index())
print(s4)
print('-' * 50)

# s4.loc[0: 3]  loc索引值不唯一时也报错
# 按索引值下标取值 iloc：左闭右开；loc：左闭右闭
print(s4.iloc[0: 3])
s4[0: 3]        # 默认用的位置索引  s4[0: 3] -> s4.iloc[0: 3]
```

```python
# 区间[0, 8)之间随机生成4个整数
print(np.random.randint(8, size= 4))
```

```python
# s4.loc[1: 2]      loc索引值唯一时可以切片
```

```python
# DataFrame
df4 = pd.DataFrame(np.random.randn(5, 5), index=np.random.randint(5, size=5), columns=np.random.randint(5, size=5))
print(df4)
print('-' * 100)
# 轴0是行索引排序
df4_isort = df4.sort_index(axis=0, ascending=True)
print(df4_isort)
```

```python
# 轴1是列索引排序
df4_isort = df4.sort_index(axis=1, ascending=True)
print(df4_isort)
```

# 6.5 按值排序（机器学习、深度学习不重要，数据分析才重要）

```python
# 按值排序，by 后是 column 的值
import random
l = [random.randint(0, 100) for i in range(24)]         # 生成24个随机数
df4 = pd.DataFrame(np.array(l).reshape(6, 4))           # 生成6行4列的DataFrame
print(df4)
print('-' * 50)

# 给轴0重新排序，根据 by=3 即下标为3的列的数据按从大到小的顺序排序
df4_vsort = df4.sort_values(by=3, axis=0, ascending=False)      # 寻找的是columns里的3，重要
print(df4_vsort)
```

```python
# 按轴1排序，by后行索引名，交换的是列
df4_vsort = df4.sort_values(by=3, axis=1, ascending=False)      # 寻找的是index里的3
print(df4_vsort)
```

# 6.6 处理缺失数据（重要）

```python
df_data = pd.DataFrame([np.random.randn(3), [1., 2., np.nan], [np.nan, 4., np.nan], [1., 2., 3.]])
print(df_data.head())
```

```python
df_data.iloc[2, 0]
```

```python
# isnull 来判断是否有空的数据
print(df_data.isnull())
```

```python
# .sum()：默认axis=0，按列求和，True=1，False=0；结果：每一列缺失值总个数（Series，索引=列名）
print(df_data.isnull().sum())
print('-' * 50)
# len(df_data)：返回DataFrame的总行数
print(len(df_data))
print('-' * 50)
print(df_data.isnull().sum() / len(df_data))
```

# 删除缺失数据

```python
# 删除第0列含有缺失值的所有行    dropna()：删除存在缺失值的行；subset=[0]：只检查第0列（列位置为0），其他列有没有缺失不管
# subset里面写数字代表列位置；如果你的列有名字，更常用列名字符串
print(df_data.dropna(subset=[0]))
df_data
```

```python
print(df_data.dropna(axis=1))           # 某列有nan就删除该列
```

 # 填充缺失数据

```python
# 给0列的空值填为-100，按特征（按列）去填充
print(df_data.iloc[:, 0].fillna(-100.))
df_data
```

```python
for i in df_data.columns:
    print(df_data.loc[:, i])
    print('-' * 50)
```

```python
# 用均值填充空值
df_data.iloc[:, 2] = df_data.iloc[:, 2].fillna(df_data.iloc[:, 2].mean())
```
