# 数据连接

```python
import pandas as pd
import numpy as np

df_obj1 = pd.DataFrame({'key': ['b', 'b', 'a', 'c', 'a', 'a', 'b'], 
                        'data1': np.random.randint(0, 10, 7)})
df_obj2 = pd.DataFrame({'key': ['a', 'b', 'd'], 
                        'data2': np.random.randint(0, 10, 3)})

print(df_obj1)
print('-' * 50)
print(df_obj2)
print(type(df_obj2))                # <class 'pandas.DataFrame'>
```

```python
# 默认连接使用相同的列名，连接方式是内连接（只保留两边都匹配的数据）
pd.merge(df_obj1, df_obj2)
```

```python
# 把左右索引作连接键join，因为df_obj2只有三行数据，故index索引只有三行
pd.merge(df_obj1, df_obj2, left_index=True, right_index=True)
```

```python
# 内连接，保留两边都匹配的数据
pd.merge(df_obj1, df_obj2, on='key')
```

```python
df_obj1 = df_obj1.rename(columns={'key': 'key1'})
df_obj2 = df_obj2.rename(columns={'key': 'key2'})
```

```python
df_obj2
```

```python
# 两张表连接键名字不一样时使用：left_on='key1'：左表df_obj1拿key1这一列作为连接键；right_on='key2'：右表df_obj2拿key2这一列作为连接键
# 如果两表键同名，直接写 on='key'；名字不同，必须分开写 left_on/right_on
# 保留两边都匹配的数据
pd.merge(df_obj1, df_obj2, left_on='key1', right_on='key2')
```

```python
# 全外连接，取左表 key1 集合 ∪ 右表 key2 集合（并集）
# 1. 两边都能匹配上的行：正常合并，数值都保留
# 2. 只在左表存在，右表无匹配：右表字段填充NaN
# 3. 只在右表存在，左表无匹配：左表字段填充NaN
pd.merge(df_obj1, df_obj2, left_on='key1', right_on='key2', how='outer')
```

```python
# 左外连接（只关心左表全部数据，丢失右表独有的行）
pd.merge(df_obj1, df_obj2, left_on='key1', right_on='key2', how='left')
```

```python
# 右外连接（不一定右表只有三行所以连接后就只有三行）
pd.merge(df_obj1, df_obj2, left_on='key1', right_on='key2', how='right')
```

```python
# np.random.randint(low, high, size)：low：下界（包含）；high：上界（不包含）；size：生成多少个数字
df_obj1 = pd.DataFrame({'key': ['b', 'b', 'a', 'c', 'a', 'a', 'b'], 
                        'data': np.random.randint(0, 10, 7)})               # 左闭右开
df_obj2 = pd.DataFrame({'key': ['a', 'b', 'd'],
                        'data': np.random.randint(0, 10, 3)})
# 给相同的数据列添加后缀
print(pd.merge(df_obj1, df_obj2, on='key', suffixes=('_left', '_right'))) 
```

```python
# 按索引连接
df_obj1 = pd.DataFrame({'key': ['b', 'b', 'a', 'c', 'a', 'a', 'b'],
                        'data1': np.random.randint(0, 10, 7)})
df_obj2 = pd.DataFrame({'data2': np.random.randint(0, 10, 3)}, index=['a', 'b', 'd'])
print(df_obj1)
print('-' * 50)
print(df_obj2)
```

```python
# left_on='key'：左表df_obj1 -> 使用普通列 key 作为连接键
# right_index=True：右表 df_obj2 -> 使用行索引作为连接键
# 默认 how='inner'（内连接）
# 以df1为主，拿df1.key去匹配df2的索引
print(pd.merge(df_obj1, df_obj2, left_on='key', right_index=True))
```

```python
# 以df2为主，拿df2的索引去匹配df1.key，所以前三行都为a
pd.merge(df_obj2, df_obj1, left_index=True, right_on='key')
```
